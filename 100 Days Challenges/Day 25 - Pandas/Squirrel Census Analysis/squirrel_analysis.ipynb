```python
import pandas as pd

squirrels_data = pd.read_csv('2018_Central_Park_Squirrel_Census_-_Squirrel_Data.csv')
squirrels_df = pd.DataFrame(squirrels_data)
squirrels_df.head()
```

```python
squirrels_df.info()
```

## My Methods

```python
# using the value_counts method which returns an array. bad for indexing, but if you need quick data it's the way to go.

primary_colors = squirrels_df['Primary Fur Color'].value_counts()
primary_colors.to_csv('primary_color.csv') # jupyter notebook already takes the relative path so all you need to do is name the file
primary_colors
```

```python
# a cleaner way is to use the group_by method which gives you more control and indexing. useful for big data.

primary_colors_df = squirrels_df.groupby('Primary Fur Color', as_index= False).size()
primary_colors_df.to_csv('primary_color_df.csv')
primary_colors_df
```
